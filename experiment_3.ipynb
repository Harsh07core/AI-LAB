```
goal =(1,2,3,4,5,6,7,8,0)
def manhattan(state):
  dist = 0
  for i in range(9):
    if state[i] !=0:
      goal_pos = goal.index(state[i])
      dist += abs(i%3 - goal_pos%3) + abs(i//3 - goal_pos//3)
  return dist
```
