```
cities = [(0,0),(1,3),(4,3),(6,1)]
def distance(a,b):
    return math.sqrt((a[0]-b[0])**2+(a[1]-b[1])**2)
```

```
def total_cost(path):
  cost = 0
  for i in range(len(path)-1):
    cost += distance(path[i], path[i+1])
  cost += distance(path[-1], path[0])
  return cost
```
