```
from collections import deque

# Generate valid moves
def get_moves(state):
    rods = [list(r) for r in state]
    next_states = []

    for i in range(3):
        if rods[i]:
            for j in range(3):
                if i != j:
                    if not rods[j] or rods[i][-1] < rods[j][-1]:
                        new_rods = [list(r) for r in state]
                        disk = new_rods[i].pop()
                        new_rods[j].append(disk)
                        next_states.append(tuple(tuple(r) for r in new_rods))
    return next_states


# ---------------- DFS ----------------
def dfs(state, goal, visited):
    if state == goal:
        return [state]

    visited.add(state)

    for move in get_moves(state):
        if move not in visited:
            path = dfs(move, goal, visited)
            if path:
                return [state] + path
    return None


# ---------------- BFS ----------------
def bfs(start, goal):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        state = path[-1]

        if state == goal:
            return path

        if state not in visited:
            visited.add(state)
            for move in get_moves(state):
                queue.append(path + [move])


# ---------------- IDDFS ----------------
def dls(state, goal, depth, visited):
    if state == goal:
        return [state]
    if depth == 0:
        return None

    visited.add(state)

    for move in get_moves(state):
        if move not in visited:
            path = dls(move, goal, depth-1, visited)
            if path:
                return [state] + path
    return None


def iddfs(start, goal):
    depth = 0
    while True:
        visited = set()
        path = dls(start, goal, depth, visited)
        if path:
            return path
        depth += 1


# ---------------- MAIN ----------------
n = 3
start = (tuple(range(n,0,-1)), (), ())
goal = ((), (), tuple(range(n,0,-1)))

print("DFS Solution:")
print(dfs(start, goal, set()))

print("\nBFS Solution:")
print(bfs(start, goal))

print("\nIDDFS Solution:")
print(iddfs(start, goal))
```
